Un projet Deep Learning ne commence pas par le choix d’une couche ou d’une commande. Il commence par une **question** : quelle réponse veut-on obtenir à partir de quelles données ? Les données sont ensuite représentées par des tenseurs, traversent un réseau et produisent une prédiction.

Pendant l’entraînement, la prédiction est comparée à la bonne réponse par une **fonction de perte**, ou *loss*. La rétropropagation calcule comment chaque poids a contribué à l’erreur, puis l’optimiseur modifie légèrement ces poids. Cette boucle est répétée sur de nombreux exemples.

## Architecture et framework : deux choix différents

![Différence entre forme des données, architecture et sortie](../ressources/illustrations/jour_02/00_architectures_et_frameworks.png)

Une **architecture** décrit l’organisation des couches et la manière dont l’information circule. Un **framework** est l’outil logiciel utilisé pour écrire, entraîner et exécuter cette architecture.

| Nom | Nature | Idée principale |
|---|---|---|
| Réseau Dense | architecture | chaque neurone reçoit toutes les sorties de la couche précédente |
| CNN | architecture | exploite la proximité spatiale, notamment dans les images |
| RNN, LSTM, GRU | architectures séquentielles | conserve un état pour traiter des éléments ordonnés |
| Autoencodeur | organisation encodeur-décodeur | apprend à reconstruire ses entrées |
| TensorFlow | framework | fournit calcul automatique, entraînement et déploiement de modèles |
| Keras | interface de haut niveau | simplifie ici la construction de modèles TensorFlow |
| PyTorch | framework | exprime les mêmes principes avec une boucle souvent plus explicite |

Un CNN n’est donc pas un concurrent de TensorFlow ou de PyTorch. Le CNN est le plan du réseau ; TensorFlow/Keras ou PyTorch sont les outils permettant de le construire. Dans ce cours, **TensorFlow/Keras sert de fil conducteur**. PyTorch sera ensuite présenté en traduisant un réseau déjà compris.

## De l’image visible à la classe prédite

![Pipeline de classification d’une image](../ressources/illustrations/jour_02/00_pipeline_classification_image.png)

Un réseau ne reçoit pas directement l’idée d’un chiffre. Il reçoit un tableau de nombres. Ses couches transforment progressivement ce tableau en une sortie numérique.

Pendant l’entraînement, le **label** sert à mesurer l’écart entre la sortie et la bonne réponse, puis à corriger les poids. Pendant l’inférence, le label n’est pas disponible : le réseau reçoit uniquement une nouvelle image.

La **classification** attribue une classe à l’image entière. D’autres tâches existent : la détection cherche plusieurs objets et leurs positions, la segmentation attribue une classe à chaque pixel, et la génération produit de nouvelles données. Elles utilisent des architectures et des sorties différentes.


![Image huit par huit représentée par des nombres](../ressources/illustrations/jour_02/00_image_tenseur.png)

*Chaque pixel est une valeur ; l’ensemble organisé de ces valeurs forme un tenseur.*


![Axes et dimensions de tenseurs d’images](../ressources/illustrations/jour_02/00_tenseur_axes.png)

### Dimension, canal et batch

Une **dimension**, ou axe, décrit une direction du tableau. Dans une image en niveaux de gris, les deux axes visibles sont la hauteur et la largeur. Une image couleur ajoute généralement un axe de trois canaux : rouge, vert et bleu.

Un **batch** est un petit lot d’images traité ensemble. Sa première dimension indique le nombre d’exemples. La forme `(32, 224, 224, 3)` signifie donc : 32 images, 224 lignes de pixels, 224 colonnes et 3 canaux.

La forme est un contrat entre les données et le réseau. Une couche prévue pour `(8, 8, 1)` ne peut pas recevoir directement une image `(224, 224, 3)` sans adaptation.


## Le vocabulaire d'une image

- Un **pixel** est une petite case de l'image contenant une intensité ou une couleur.
- Un **tenseur** est un tableau de nombres pouvant avoir plusieurs dimensions.
- La **shape**, ou forme, décrit la taille de chaque dimension.
- Une image en niveaux de gris possède généralement un canal ; une image couleur possède souvent trois canaux : rouge, vert et bleu.
- Le **label** est la réponse attendue, par exemple le chiffre représenté.

Une seule image couleur de 640 × 640 pixels peut donc avoir la forme `(640, 640, 3)`. Un lot de 16 images ajoute une dimension et prend la forme `(16, 640, 640, 3)`.


## Le jeu `digits` utilisé dans les démonstrations

`load_digits()` fournit 1 797 petites images de chiffres manuscrits. Chaque image mesure 8 × 8 pixels et chaque pixel possède une intensité comprise entre 0 et 16.

`digits.images` contient toutes les images. Sa forme possède donc trois axes : nombre d’images, hauteur et largeur. `digits.target` contient un entier de 0 à 9 pour chaque image.

Dans la boucle d’affichage, `zip` associe un axe graphique, une image et son label. `imshow` transforme la matrice en image visible ; `cmap="gray_r"` choisit une palette en niveaux de gris. Le graphique ne change pas les données, il les rend seulement lisibles pour nous.


```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_digits


digits = load_digits()
images = digits.images
labels = digits.target
print("Toutes les images :", images.shape)
print("Une image          :", images[0].shape)
print("Label associé      :", labels[0])

fig, axes = plt.subplots(1, 5, figsize=(10, 2.2))
for ax, image_value, label in zip(axes, images[:5], labels[:5]):
    ax.imshow(image_value, cmap="gray_r")
    ax.set_title(f"label = {label}")
    ax.axis("off")
plt.tight_layout()
plt.show()
```

## Pourquoi normaliser les pixels ?

Les pixels du jeu `digits` vont de 0 à 16. Les diviser par 16 les place entre 0 et 1. Cette **normalisation** facilite généralement l’optimisation : les calculs internes manipulent des valeurs sur une échelle régulière.

Normaliser ne change ni la forme visible ni le label. Cela change seulement la représentation numérique fournie au réseau. Il faut appliquer exactement la même règle au train, à la validation, au test et aux futures images.

Il ne faut pas confondre normalisation et redimensionnement. Normaliser change les valeurs des pixels ; redimensionner change la hauteur et la largeur. Convertir une image couleur en gris change encore autre chose : le nombre de canaux.

### Type numérique et mémoire

Les images sont converties en `float32`, un format décimal courant en Deep Learning. Un mauvais type peut empêcher une division correcte ou consommer inutilement de la mémoire. Vérifier `shape`, minimum, maximum et type numérique constitue donc un premier contrôle systématique.
