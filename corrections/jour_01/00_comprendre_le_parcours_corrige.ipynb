## Pourquoi commencer par les fondamentaux ?

Le Machine Learning permet à un programme d'ajuster un comportement à partir d'exemples plutôt que de recevoir toutes ses règles à la main. Le Deep Learning appartient au Machine Learning et utilise des réseaux de neurones comportant plusieurs couches.

Pendant ces deux jours, nous construirons un vocabulaire commun et une méthode de travail : formuler une question, organiser les données, entraîner un modèle, mesurer ses erreurs puis analyser ses limites.


## IA, Machine Learning et Deep Learning : qui contient quoi ?

![Ensembles Intelligence artificielle Machine Learning et Deep Learning](../../ressources/illustrations/jour_01/00_ia_ml_dl.png)

L’**intelligence artificielle**, ou IA, est le terme le plus large. Il regroupe différentes techniques capables d’automatiser une tâche que l’on associe habituellement à l’intelligence : raisonner avec des règles, rechercher une solution ou apprendre à partir d’exemples.

Le **Machine Learning** est une partie de l’IA. Au lieu d’écrire manuellement une règle pour chaque situation, on fournit des exemples à un algorithme afin qu’il ajuste un modèle.

Le **Deep Learning** est lui-même une partie du Machine Learning. Il utilise des réseaux de neurones comportant plusieurs couches. Il est particulièrement utile avec des données complexes comme les images, le son ou le texte, mais il n’est pas automatiquement meilleur sur tous les problèmes.

Ces termes décrivent des familles de méthodes. Ils ne signifient pas que la machine pense, comprend ou possède une intention.


![Parcours allant des données à la production](../../ressources/illustrations/jour_01/00_parcours_ml_dl.png)

*Le Machine Learning et le Deep Learning partagent une même exigence : apprendre sur des données puis évaluer honnêtement les erreurs.*


## Écrire des règles ou apprendre à partir d’exemples

![Comparaison programmation classique et Machine Learning](../../ressources/illustrations/jour_01/00_programmation_vs_ml.png)

Dans un programme classique, le développeur écrit les règles puis les applique aux données. En Machine Learning, on fournit des **exemples historiques** et leurs bonnes réponses ; l’algorithme cherche alors des relations utiles.

L’ensemble des exemples s’appelle un **jeu de données**, ou *dataset*. Une réponse connue associée à un exemple est souvent appelée **label**. Dans notre cas, le label indique si une panne est survenue dans les sept jours.

Pendant l’entraînement, le modèle ajuste ses **paramètres**. Ce sont les valeurs internes apprises, par exemple les coefficients d’une régression ou les poids d’un réseau. Les **hyperparamètres** sont au contraire choisis avant l’entraînement : profondeur maximale d’un arbre, nombre d’arbres ou nombre d’époques.

Apprendre ne signifie pas mémoriser parfaitement le passé. Le but est de **généraliser**, c’est-à-dire rester utile sur de nouvelles observations qui ressemblent suffisamment aux situations apprises.


## La carte du Machine Learning

![Panorama des familles, étapes et pistes du Machine Learning](../../ressources/illustrations/jour_01/00_panorama_machine_learning.png)

Le **Machine Learning** ne désigne pas un seul algorithme. C'est un ensemble de façons d'apprendre des relations à partir de données.

- En **apprentissage supervisé**, chaque exemple possède une réponse connue. On peut prédire une catégorie, avec la classification, ou une valeur numérique, avec la régression.
- En **apprentissage non supervisé**, aucune réponse n'est fournie. On cherche notamment des groupes, des représentations plus compactes ou des observations inhabituelles.
- En **apprentissage par renforcement**, un agent apprend en essayant des actions et en recevant des récompenses ou des pénalités.

Quel que soit le type choisi, les données, l'évaluation et le suivi du modèle restent essentiels. Durant ces deux jours, nous pratiquons surtout l'apprentissage supervisé. Les autres branches sont montrées pour vous permettre de situer ce que vous pourrez approfondir ensuite.


## Quatre mots à ne pas confondre

- Un **algorithme** est une méthode générale d'apprentissage, par exemple une régression logistique ou un arbre de décision.
- Un **modèle** est le résultat obtenu après avoir appliqué cet algorithme à des données précises.
- L'**entraînement**, souvent lancé par `fit`, ajuste les paramètres du modèle à partir d'exemples connus.
- L'**inférence**, souvent lancée par `predict`, utilise le modèle entraîné sur une nouvelle observation sans réapprendre.

Exemple : la recette d'un gâteau ressemble à l'algorithme ; le gâteau réellement préparé ressemble au modèle. Deux jeux de données différents peuvent produire deux modèles différents avec le même algorithme.

## Le fil rouge

Nous disposerons de mesures décrivant des machines : température, vibration, pression, charge, âge et temps depuis la maintenance. Pour chaque ligne historique, nous savons si une panne est survenue dans les sept jours.

Notre première mission sera donc : **à partir des informations disponibles maintenant, produire une estimation du risque de panne prochaine**.


## Première rencontre avec les données

La cellule suivante charge le fichier CSV dans un `DataFrame` pandas. Un **CSV** est un fichier tabulaire : chaque ligne représente une observation et chaque colonne une information.

- `len(df)` compte les observations ;
- `df.shape[1]` donne le nombre de colonnes ;
- `value_counts()` compte les exemples de chaque classe ;
- `head(3)` affiche les trois premières lignes sans modifier les données.

Les bibliothèques importées ont chacune un rôle : **pandas** manipule les tableaux, **NumPy** fournit des calculs numériques et **Matplotlib** construit des graphiques. `Path` permet de retrouver le fichier quel que soit le dossier depuis lequel le notebook est lancé.


```python
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt


def find_course_root() -> Path:
    for candidate in (Path.cwd(), *Path.cwd().parents):
        if (candidate / "ressources" / "donnees" / "maintenance_machines.csv").exists():
            return candidate
    raise FileNotFoundError("Racine du cours introuvable. Ouvrez le notebook depuis le dossier du cours.")


ROOT = find_course_root()
DATA_FILE = ROOT / "ressources" / "donnees" / "maintenance_machines.csv"
plt.style.use("seaborn-v0_8-whitegrid")

df = pd.read_csv(DATA_FILE)
print(f"{len(df)} observations et {df.shape[1]} colonnes")
print(df["failure_7d"].value_counts().rename(index={0: "pas de panne", 1: "panne"}))
df.head(3)
```

## Comment lire cette première sortie ?

Une ligne ne représente pas « toute une machine », mais une machine observée à un instant de référence. Les colonnes décrivent cette observation. La colonne `failure_7d` vaut `1` lorsqu’une panne survient dans les sept jours et `0` sinon.

Le comptage montre que les pannes sont rares. Ce **déséquilibre des classes** sera important : obtenir beaucoup de bonnes réponses sur les cas normaux ne garantit pas que les pannes soient correctement détectées.

À ce stade, aucun modèle n’a encore été entraîné. Nous vérifions seulement ce que contient le jeu de données et quelle question il permet de poser.


## Un modèle ne « comprend » pas une machine

Il repère des régularités statistiques dans les exemples présentés. Une vibration élevée peut être associée aux pannes dans notre historique sans être leur cause unique. Le modèle ne remplace donc ni le diagnostic technique, ni la connaissance du procédé.
